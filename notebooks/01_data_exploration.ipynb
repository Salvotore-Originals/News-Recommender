```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd().parent

MIND_TRAIN = ROOT / "data" / "raw" / "mind" / "train"

news_path = MIND_TRAIN / "news.tsv"

news = pd.read_csv(
    news_path,
    sep="\t",
    header=None
)

print("Shape:", news.shape)

news.columns = [
    "article_id",
    "category",
    "subcategory",
    "title",
    "abstract",
    "url",
    "title_entities",
    "abstract_entities"
]

print(news.info())
print(news.isna().sum())
print("Duplicate IDs:", news["article_id"].duplicated().sum())
print("Unique categories:", news["category"].nunique())
print(news["category"].value_counts().head(20))
```

```python
behaviors_path = MIND_TRAIN / "behaviors.tsv"

behaviors = pd.read_csv(
    behaviors_path,
    sep="\t",
    header=None
)

print(behaviors.shape)
print(behaviors.head())
```

```python
print("NEWS SHAPE:", news.shape)
print("BEHAVIORS SHAPE:", behaviors.shape)

print("\nNEWS:")
print(news.head(3).to_string())

print("\nBEHAVIORS:")
print(behaviors.head(3).to_string())

print("\nTIME RANGE:")
print(behaviors["timestamp"].min())
print(behaviors["timestamp"].max())
```
